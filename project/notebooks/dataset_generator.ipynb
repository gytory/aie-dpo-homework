```python
display(Markdown("# Статистика по ответам"))

df_docs['char_count'] = df_docs['response_text'].apply(lambda x: len(str(x)))

length_stats_docs = pd.DataFrame({
    'Показатель': ['Максимальная', 'Минимальная', 'Средняя'],
    'Количество символов': [
        df_docs['char_count'].max(),
        df_docs['char_count'].min(),
        round(df_docs['char_count'].mean(), 1)
    ]
})
display(length_stats_docs)
display(Markdown("Поскольку ответы техподдержки и так достаточно короткие (средняя длина окло 240 символов) - чанкинг здесь не требуется."))

display(Markdown("## Распределение ответов по темам"))

topic_stats = df_docs['topic'].value_counts().reset_index()
topic_stats.columns = ['Тема', 'Количество ответов']
display(topic_stats)
display(Markdown("Распределение ответов по темам равномерно (по 8 на каждую).\n\n"
                 "Это полезно для разделения смысла ответов и распределения вопрос-ответ логичным образом."))
```

```python
import re
from collections import Counter
display(Markdown("# Статистика по запросам"))
df_queries['char_count'] = df_queries['query_text'].apply(lambda x: len(str(x)))
df_queries['word_count'] = df_queries['query_text'].apply(lambda x: len(str(x).split()))

basic_stats = pd.DataFrame([{
    'Средняя длина (символов)': round(df_queries['char_count'].mean(), 1),
    'Средняя длина (слов)': round(df_queries['word_count'].mean(), 1),
    'Минимум слов в запросе': df_queries['word_count'].min(),
    'Максимум слов в запросе': df_queries['word_count'].max()
}])
display(basic_stats)

display(Markdown("## Часто встречающиеся ключевые слова"))

stop_words = {'и', 'в', 'на', 'с', 'по', 'к', 'у', 'о', 'об', 'от', 'до', 'за', 'из',
              'не', 'нет', 'был', 'была', 'было', 'были', 'этот', 'эта', 'это', 'эти',
              'весь', 'вся', 'все', 'всё', 'такой', 'такая', 'такое', 'такие',
              'который', 'которая', 'которое', 'которые', 'уже', 'еще', 'вот', 'там',
              'тут', 'здесь', 'тогда', 'потом', 'потому', 'поэтому', 'очень', 'можно',
              'нужно', 'надо', 'будет', 'будьте', 'пожалуйста', 'здравствуйте', 'добрый',
              'день', 'вечер', 'утро', 'привет', 'спасибо', 'помогите', 'подскажите',
              'скажите', 'что', 'как', 'когда', 'где', 'почему', 'зачем', 'куда',
              'через', 'для', 'после', 'сколько', 'при', 'пишет', 'хотя', 'могу', 'делать',
              'работает', 'открывается', 'уважаемая', 'уважаемые', 'поддержка'}

all_words = []
for query in df_queries['query_text']:
    cleaned = re.sub(r'[^а-яё\s]', '', str(query).lower())
    words = cleaned.split()
    for word in words:
        if len(word) > 2 and word not in stop_words:
            all_words.append(word)

word_freq = Counter(all_words)
top_words = pd.DataFrame(sorted(word_freq.most_common(25), key=lambda x: x[1], reverse=True), columns=['Ключевое слово', 'Частота'])
display(top_words)
display(Markdown("Приведены самые частые слова в запросах, которые являются основной темой запросов."))
```
